# Partie 1 : Listes Python vs tableaux NumPy

## Objectif de la partie
Avant de plonger dans les fonctionnalités avancées de NumPy, nous devons comprendre **pourquoi** nous utilisons cette bibliothèque plutôt que les listes classiques de Python.
## Ce qui est attendu
Nous disposons des températures mesurées par un capteur : 24.5, 25.1, 26.3, 27.0 et 25.8.
1. Nous allons créer une **liste Python** classique pour stocker ces valeurs.
2. Nous allons créer un **tableau NumPy** équivalent.
3. Nous allons multiplier ces deux collections par 2.
4. Nous observerons et noterons la **différence fondamentale** dans le comportement

```python
%pip install numpy
```

```python
import numpy as np

#Création de liste python et du tableau numpy
temp_py = [24.5, 25.1, 26.3, 27.0, 25.8]
temp_nump = np.array(temp_py)
print("liste_python:" , temp_py)
print("numpy_array:" , temp_nump)

#Multiplication par 2
mult_py = temp_py * 2
mult_nump = temp_nump * 2
print( "liste_python*2:" , mult_py)
print("numpy_array*2:" , mult_nump)
```

Nous observons et notons la **différence fondamentale** dans le comportement :
   - La liste Python va dupliquer les éléments lors de la multiplication par 2(elle s'allonge).
   - Le tableau NumPy va multiplier mathématiquement chaque valeur de la liste par 2 (calcul vectoriel).

# Partie 5 – Filtrage booléen

## Objectif de la partie
L'indexation que nous avons vue dans la Partie 4 consiste à extraire des données **par position** (les 3 premières, la 8ème, etc.). Mais dans la vraie vie, on veut souvent extraire des données **selon une condition** : "Donne-moi toutes les températures trop élevées", ou "Donne-moi toutes les températures normales".
Pour cela, NumPy utilise une technique très puissante appelée **filtrage booléen**. On crée un tableau de Vrai/Faux (booléens), puis on l'utilise comme "masque" pour ne garder que les valeurs Vrai.
## Ce qui est attendu
À partir du tableau des 24 températures d'une journée (créé dans la Partie 3), nous allons :
1. Afficher les températures **supérieures à 30 °C**.
2. Afficher les températures **"normales"** c'est-à-dire comprises **entre 20 °C et 30 °C** (inclus).
3. Supposons qu'une température supérieure à 35 °C soit **anormale**. Créer un tableau de booléens nommé `anomalies` qui précise, pour chaque heure, si la température est anormale (True) ou non (False).
4. Déterminer, à l'aide de `anomalies`, le **nombre** de températures anormales.

```python
import numpy as np

# Partie 5: Filtrage booléen
# 1. Afficher les températures supérieures à 30 °C
np.random.seed (42)
temp_journée= np.random.uniform (20, 30, 24)
print ("temp_journée: ", temp_journée)
# Affichage du masque de tri
masque_over_30 = temp_journée > 30
temp_over_30 = temp_journée[masque_over_30] 
print ("1. Températures > 30°C :", temp_over_30)

# 2. Afficher les températures « normales » càd comprises entre 20 °C et 30 °C

condition_min = temp_journée >= 20
condition_max = temp_journée <= 30
masque_normal = condition_min & condition_max
temp_normales = temp_journée[masque_normal]
print("2. Températures normales (20-30) :", temp_normales)
# 3. Création du tableau des anomalies (Température > 35°C)
# On stocke le résultat (Vrai/Faux) dans une variable appelée "anomalies".
anomalies = temp_journée > 35
table_anomalies = temp_journée [anomalies]
print("3. Tableau des anomalies:", table_anomalies)
# 4. Détermination du nombre de température anormales
nombre_anomalies = np.sum(table_anomalies)
print("4. Nombre de températures anormales (>35°C) :", int(nombre_anomalies))
```

# Partie 6 – Manipulation des dimensions

## Objectif de la partie
Jusqu'à présent, nous avons travaillé avec des tableaux à une seule dimension (des vecteurs). Mais dans la vraie vie, les données des capteurs sont souvent organisées en **matrices** (2 dimensions) : les lignes représentent les capteurs, et les colonnes représentent les heures.
Dans cette partie, nous allons :
- Créer une matrice 2D.
- Apprendre à changer sa forme (l'aplatir, la transposer).
- Extraire des lignes (un capteur spécifique).
- Extraire des colonnes (une heure spécifique).
- Extraire une sous-matrice (un groupe de capteurs sur une plage horaire précise).
## Ce qui est attendu
On suppose qu'on a 5 capteurs.

1. Créer aléatoirement la matrice représentant les températures mesurées par ces capteurs durant une journée (5 capteurs × 24 heures).
2. Transformer la matrice en un tableau unidimensionnel (aplatir).
3. Déterminer la transposée de la matrice.
4. Extraire toutes les mesures du capteur 1.
5. Extraire toutes les mesures du capteur 5.
6. Extraire les mesures de tous les capteurs à 10h.
7. Extraire les mesures des capteurs 2 à 4 entre 8h et 12h.

```python
import numpy as np
# 1. Créer aléatoirement la matrice représentant les températures mesurées par ces capteurs durant une journée
np.random.seed(42)
matrice_temp = np.round(np.random.uniform(20, 30, size= (5,24)), 2)
print ("La matrice de température des 5 capteurs est: ", matrice_temp)
# 2. Transformer la matrice en un tableau unidimensionnel
table_unidim = matrice_temp.flatten()
print("La table unidimensionnelle est: ", table_unidim)
# 3. Déterminer la transposée de la matrice
table_transpose = matrice_temp.T
print("La Table transposée est: ", table_transpose)
# 4. Extraire toutes les mesures du capteur 1
capteur_1 = matrice_temp [0]
print("Le capteur 1 affiche: ",capteur_1)
# 5. Extraire toutes les mesures du capteur 5
capteur_5 = matrice_temp [4]
print("Le capteur 5 affiche: ",capteur_5)
# 6. Extraire les mesures de tous les capteurs à 10H
all_capteurs_10H = matrice_temp [:, 10]
print("Les capteurs affichent à 10H: ", all_capteurs_10H)
# 7. Extraire les mesures des capteurs 2 à 4 entre 8h et 12h.
capteur2_4 = matrice_temp[1:4, 8:12]
print("Les mesures des capteurs 2 à 4 entre 8H et 12H sont: ", capteur2_4)

```

# Partie 7 – Concaténation
## Objectif de la partie
Dans la Partie 6, nous avions 5 capteurs dans un premier bâtiment. Maintenant, on ajoute un **second bâtiment** avec 3 capteurs supplémentaires. Notre objectif est de **fusionner** ces deux ensembles de données en une seule grande matrice qui contient les 8 capteurs.
La concaténation est l'opération qui consiste à "coller" deux tableaux ensemble, soit verticalement (en ajoutant des lignes), soit horizontalement (en ajoutant des colonnes).

```python
import numpy as np
# 1. Créer aléatoirement la matrice représentant les températures mesurées par les 3 capteurs durant une journée
np.random.seed(42)
matrice_temp = np.round(np.random.uniform(20, 30, size= (5,24)), 2)
print ("La matrice de température des 5 capteurs est: ", matrice_temp)

np.random.seed(42)
matrice_temp_bat2 = np.round(np.random.uniform(20, 30, size= (3,24)), 2)
print ("La matrice de température des 3 capteurs est: \n", matrice_temp_bat2)
# Déterminer de deux manières différentes la matrice qui fusionne les températures mesurées par les 8 capteurs durant une journée
# Concaténation par np.concatenate
table_concat_1 = np.concatenate([matrice_temp, matrice_temp_bat2], axis=0)
print ("Le tableau concaténer est : \n", table_concat_1)
# Concaténation avec np.vstack
table_contact_2= np.vstack((matrice_temp, matrice_temp_bat2))
print("Le tableau concaténer avec np.vstack est: \n", table_contact_2)  
```

# Partie 8 – Calcul scientifique

## Objectif de la partie
Maintenant que nous savons manipuler et extraire des données, nous allons utiliser les fonctions mathématiques de NumPy pour analyser les températures. Ces calculs (moyenne, écart-type, normalisation) sont le **cœur de la data science** : ils permettent de résumer l'information et de préparer les données pour les algorithmes de Machine Learning.

```python
import numpy as np
# 1. Pour les températures calculer le minimum, le maximum, la somme, la moyenne, la médiane, la variance et l’écart-type
np.random.seed(42)
temp_journée = np.round(np.random.uniform(20, 30, 24), 2)
print (temp_journée)
# Le minimum 
min_temp_journée = np.min(temp_journée)
max_temp_journée = np.max(temp_journée)
sum_temp_journée = np.sum(temp_journée)
moy_temp_journée = np.mean(temp_journée)
med_temp_journée = np.median(temp_journée)
var_temp_journée = np.var(temp_journée)
std_temp_journée = np.std(temp_journée)
print("La température minimale dans la journée est: ", min_temp_journée)
print("La température maximale dans la journée est: ", max_temp_journée)
print("La somme des températures dans la journée est: ", sum_temp_journée)
print("La moyenne des températures dans la journée est: ", moy_temp_journée)
print("La médiane des températures dans la journée est: ", med_temp_journée)
print("La variance des températures dans la journée est: ", var_temp_journée)
print("L'écart-type des températures dans la journée est: ", std_temp_journée)

# 2. Standardiser le tableau des températures puis vérifier les valeurs de la moyenne et de l’écarttype du tableau standardisé
temp_standardisee = (temp_journée - moy_temp_journée) / std_temp_journée
print ("Le tableau standarisée est: ", temp_standardisee)

# Vérification
moyenne_std = np.round(np.mean(temp_standardisee))
print ("La moyenne standarisée est: ",abs(moyenne_std))
ecart_std = np.std(temp_standardisee)
print ("L'écart-type standarisée est: ",ecart_std)
# 3. Normaliser (Min-Max) le tableau des températures puis vérifier les valeurs du tableau normalisé
temp_normalisée = (temp_journée - min_temp_journée) / (max_temp_journée - min_temp_journée)
print ("La température normalisée est: ", temp_normalisée)

# Vérification
min_norm = np.min(temp_normalisée)
print (" Le Minimum du tableau normalisé est: ", min_norm)
max_norm = np.max(temp_normalisée)
print (" Le maximum du tableau normalisé est: ", min_norm)
# 4. Convertir le tableau initial en Fahrenheit
temp_fahrenheit = temp_journée * (9/5) + 32
print("Le tableau en celsuis Fareinheit est: ", temp_fahrenheit)




```

# Partie 9 – Broadcasting

## Objectif de la partie
Nous avons vu qu’on peut multiplier un tableau par un nombre (ex: `tableau * 2`) et que ce nombre s’applique à toutes les cases. C’est un cas particulier de **broadcasting** (diffusion).
Le broadcasting va plus loin : il permet d’appliquer une opération entre deux tableaux de **formes différentes**, à condition qu’ils soient "compatibles". Ici, nous allons appliquer une **correction différente pour chaque capteur** sur 3 jours de mesures.
## Ce qui est attendu
1. Créer aléatoirement la matrice `temperatures_vsd` représentant les températures mesurées un **vendredi, samedi et dimanche successifs** dans le **2nd bâtiment** (qui possède 3 capteurs). Cette matrice aura donc une forme de **(3 jours, 3 capteurs)**.
2. On suppose que les valeurs de calibration des capteurs sont stockées dans le tableau `correction = np.array([1, -0.5, 0.2])`. 
   Appliquer automatiquement ces corrections pour trouver les mesures corrigées : `mesures_corrigees = temperatures_vsd + correction`.

```python
import numpy as np

# Partie 9 – Broadcasting (Mise à jour : Lignes = Capteurs)
# 1. Créer la matrice 3x3 (3 capteurs, 3 heures)
np.random.seed(42)

temperatures_3x3 = np.round(np.random.uniform(20, 30, size=(3, 3)), 2)

print("Les températures mises à jours sont: ", temperatures_3x3)

# 2. Appliquer les corrections (Broadcasting)

correction = np.array([1, -0.5, 0.2])
mesures_corrigees = temperatures_3x3 + correction
print("Températures corrigées :", mesures_corrigees)

# Vérification
print("Différence (corrigé - original) :")
print(mesures_corrigees - temperatures_3x3)
```
