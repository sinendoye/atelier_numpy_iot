# Partie 1 : Listes Python vs tableaux NumPy

## Objectif de la partie
Avant de plonger dans les fonctionnalités avancées de NumPy, nous devons comprendre **pourquoi** nous utilisons cette bibliothèque plutôt que les listes classiques de Python.
## Ce qui est attendu
Nous disposons des températures mesurées par un capteur : 24.5, 25.1, 26.3, 27.0 et 25.8.
1. Nous allons créer une **liste Python** classique pour stocker ces valeurs.
2. Nous allons créer un **tableau NumPy** équivalent.
3. Nous allons multiplier ces deux collections par 2.
4. Nous observerons et noterons la **différence fondamentale** dans le comportement

```python
%pip install numpy
```

```python
import numpy as np

#Création de liste python et du tableau numpy
temp_py = [24.5, 25.1, 26.3, 27.0, 25.8]
temp_nump = np.array(temp_py)
print("liste_python:" , temp_py)
print("numpy_array:" , temp_nump)

#Multiplication par 2
mult_py = temp_py * 2
mult_nump = temp_nump * 2
print( "liste_python*2:" , mult_py)
print("numpy_array*2:" , mult_nump)
```

Nous observons et notons la **différence fondamentale** dans le comportement :
   - La liste Python va dupliquer les éléments lors de la multiplication par 2(elle s'allonge).
   - Le tableau NumPy va multiplier mathématiquement chaque valeur de la liste par 2 (calcul vectoriel).

# Partie 2 – Création des données

## Objectif de la partie
Maintenant que nous avons compris la différence entre une liste Python et un tableau NumPy, nous allons apprendre à **générer** des tableaux de données sans avoir à les écrire un par un. 
Dans la vraie vie (et dans l'atelier), on a souvent besoin de créer des tableaux de zéros, de nombres alignés, ou de valeurs aléatoires pour simuler des données de capteurs.
## Ce qui est attendu
Nous allons utiliser les fonctions de création de NumPy pour répondre aux 7 questions suivantes. Chaque question correspond à une situation concrète

```python
import numpy as np

#Partie 2: Création de données
# 1. Créer un tableau représentant les mesures de température de 10 capteurs, initialement à zéro
temp_initiale = np.zeros(10)
# 2. Créer un tableau contenant 10 valeurs égales à 1
table_1= np.ones(10)
# 3. Créer un tableau de 10 valeurs initialisées à 25.
table_25 = np.full(10, 25)
# 4. Créer une matrice identité de taille 5 × 5
mat_id = np.eye (5,5)
# 5. Créer les 24 heures (de 0 à 23) de mesure d’une journée
heures = np.arange(24)
# 6. Créer 10 valeurs régulièrement réparties entre 20 °C et 30 °C
temp_degrés = np.linspace(20, 30, 10)
# 7. Créer aléatoirement les températures de 100 mesures, en rendant les résultats reproductibles
np.random.seed(42)
temp_aléatoires = np.random.uniform(20, 30, 100)

#Affichage des résultats
print("1. Température initiale :", temp_initiale)
print("2. Tableau de 1s :", table_1)
print("3. Tableau de 25s :", table_25)
print("4. Matrice identité :\n", mat_id)
print("5. Heures :", heures)
print("6. Températures régulières : \n", temp_degrés)
print("7. Températures aléatoires: \n", temp_aléatoires)
```

# Partie 3 – Exploration des tableaux

## Objectif de la partie
Maintenant que nous savons créer des tableaux de données, nous devons apprendre à les **explorer** et à comprendre leur structure. Dans la vraie vie, quand on reçoit un jeu de données (un tableau NumPy), on a besoin de savoir immédiatement :
- Combien de dimensions il a (est-ce un simple vecteur, une matrice, ou un tableau à plusieurs dimensions ?)
- Combien de valeurs il contient au total.
- Quel est le type des données (nombres entiers, décimaux, texte ?).
- Quelle est sa forme (ex: 24 lignes, ou 24 valeurs sur une seule ligne ?).
Ces informations sont essentielles avant de faire le moindre calcul ou la moindre modification.
## Ce qui est attendu
Nous allons utiliser le tableau des 24 températures d'une journée, créé aléatoirement, puis nous interrogerons ses propriétés à l'aide des attributs suivants :
- `.ndim` : pour connaître le nombre de dimensions.
- `.size` : pour connaître le nombre total d'éléments.
- `.dtype` : pour connaître le type de données (int, float, etc.).
- `.shape` : pour connaître la forme du tableau (nombre de lignes, de colonnes).
L'objectif est de prendre l'habitude de toujours "tâter" les données avant de les manipuler.

```python
import numpy as np
# Partie 3: Exploration des tableaux
# 1.Créer aléatoirement un tableau représentant les 24 températures d'une journée
np.random.seed(42)
temp_journée = np.random.uniform(20, 30, 24)
# 2. Déterminer les propriétés suivantes du tableau
# dimension du tableau
dimension = temp_journée.ndim
# nombre de valeurs dans le tableau
nombre_valeurs = temp_journée.size
# type de données du tableau
type_donnees = temp_journée.dtype
# forme du tableau  
forme = temp_journée.shape

# 3. Afficher les résultats
print ("La dimension du tableau est :", dimension)
print ("Le nombre de valeurs dans le tableau est :", nombre_valeurs)
print ("Le type de données du tableau est :", type_donnees)
print ("La forme du tableau est :", forme)
```
